（2） （编程题）从键盘输入两个命题变元P和Q的真值，求他们的合取、析取、条件和双条件的真值

```python
def read_bool(name):
    while True:
        value = input(f"请输入 {name} 的真值（0=假，1=真）：").strip()
        if value in ("0", "1"):
            return value == "1"
P = read_bool("P")
Q = read_bool("Q")
print("合取:",'真'if P and Q else '假')
print("析取:",'真'if P or Q else '假')
print("条件:",'真'if (not P) or Q else '假')
print("双条件:",'真'if P == Q else '假')
```
